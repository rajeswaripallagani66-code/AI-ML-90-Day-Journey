```python
import pandas as pd
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder,MinMaxScaler
df=pd.DataFrame({'Branch':['CSE','ECE','CSE'],
                 'marks':[80,None,90],
                 'Placed':['Yes','No','Yes']})
df['marks']=SimpleImputer(strategy='mean').fit_transform(df[['marks']])
#df=pd.get_dummies(df,columns=['Branch'])
OneHotEncoder(sparse_output=False).fit_transform(df[['Branch']])
df['marks']=MinMaxScaler().fit_transform(df[['marks']])
df['Placed']=df['Placed'].map({'Yes':1,'No':0})
print(df)
```

```python
import pandas as pd
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder,MinMaxScaler,StandardScaler
df=pd.DataFrame({'Area':[1000,1200,1100],
                 'price':[50,50,500]})
Q1=df['price'].quantile(0.25)
Q3=df['price'].quantile(0.75)
IQR=Q3-Q1
df_clean=df[(df['price']>=Q1-1.5*IQR) & (df['price']<=Q3+1.5*IQR)]
print(df_clean)
```

```python
import pandas as pd
import joblib
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder,MinMaxScaler,StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer

# IMPORTANT: Create df again here
df = pd.DataFrame({'Branch':['CSE','ECE','CSE'],
                   'marks':[80,None,90],
                   'Placed':['Yes','No','Yes']})

print(df.columns) # Check - should show Branch, marks, Placed

num_features=['marks']
cat_features=['Branch']
num_pipe=Pipeline([('imputer',SimpleImputer(strategy='mean')),('scaler',StandardScaler())])
#cat_pipe=([('imputer',SimpleImputer(strategy='most_frequent')),('onehot',OneHotEncoder())])
cat_pipe = Pipeline([   # <-- Pipeline here, not []
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('onehot', OneHotEncoder())
])
full_pipe=ColumnTransformer([('num',num_pipe,num_features),('cat',cat_pipe,cat_features)])
final_data=full_pipe.fit_transform(df)
print(final_data)
joblib.dump(full_pipe,'feature_pipeline.pkl')
```
