```python
import numpy as np
import pandas as pd

# Use Numpy to create marks
np.random.seed(0)
marks = np.random.randint(50, 100, size=10) # 10 random marks
names = ['Raju','Rani','Ravi','Anu','Sai','Kiran','Geetha','Sita','Ram','Lakshman']

print(marks)
print("Mean:", np.mean(marks))
print("Max:", np.max(marks))
print("Min:", np.min(marks))
df = pd.DataFrame({'Name':names, 'Marks':marks})
print(df)
print(df.shape)
print(df.isnull().sum())
df['Branch'] = ['CSE']*5 + ['ECE']*5
print(df)
#df['Result']=df['marks'].apply(lambda x: 'Pass' if x>60 else 'Fail')
df['Result'] = df['Marks'].apply(lambda x: 'Pass' if x>60 else 'Fail')
print(df)
print(df.sort_values(by='Marks',ascending=False))
print(df.groupby('Branch')['Marks'].mean())
print(df.sort_values(by='Marks',ascending=False).head(3))
mean_val = df['Marks'].mean()
df['Marks'] = np.where(df['Marks']<60, mean_val, df['Marks'])
print(df)
mean_marks = np.mean(df['Marks'])
print(df[df['Marks'] > mean_marks + 5])
```
