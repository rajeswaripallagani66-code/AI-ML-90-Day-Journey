```python
import numpy as np
data=np.array([[80,90,70],[60,50,40],[100,95,90],[30,40,35],[70,75,80]])
print("original:\n",data)
norm=data/data.max()
print("normalied data:\n",norm)
print("avg marks per subject:\n",data.mean(axis=0))
print("total marks of all of each subject:\n ",data.sum(axis=0))
print("total of each studen:\n",data.sum(axis=1))
print("avg of each student:\n",data.mean(axis=1))
np.random.seed(0)
np.random.shuffle(data)
split=int(0.8 * len(data))
train,test=data[:split],data[split:]
print("Train:\n",train)
print("Test:\n",test)
print(len(data))
```
