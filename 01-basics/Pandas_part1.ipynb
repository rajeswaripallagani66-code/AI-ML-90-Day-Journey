```python
import pandas as pd
df=pd.DataFrame({'A':[10,20,30],'B':[40,50,60],'C':[70,80,60]},index=['x','y','z'])
print("index of x and A\n",df.loc['x','A'])
print("ilocation:\n",df.iloc[0,0])
print("index of x and z\n",df.loc['x':'z'])
print("index of 0:2\n",df.iloc[0:2])
```

```python
import pandas as pd
df=pd.DataFrame({'Branch':['CSE','CSE','ECE','ECE'],
                 'marks':[855,90,75,80],
                 'Fees':[50000,50000,40000,40000]
                 })
print(df.groupby('Branch')['marks'].mean())
print(df.groupby('Branch').agg({'marks':'mean','Fees':'sum'}))
```
