```python
import pandas as pd
df1=pd.DataFrame({'ID':[1,2,3],'Name':['Raju','Raji','Ravi']})
df2=pd.DataFrame({'ID':[1,2,3],'marks':[85,92,77]})
merged=pd.merge(df1,df2,on='ID')
print(merged)
```

```python
import pandas as pd
df_a=pd.DataFrame({'Name':['A','B'],'marks':[80,90]})
df_b=pd.DataFrame({'Name':['B','C'],'marks':[90,85]})
combined=pd.concat([df_a,df_b],ignore_index=True)
print("Before:\n",combined)
combined=combined.drop_duplicates()
print("\n After drop duplicates :\n",combined)
```

```python
import pandas as pd
df=pd.DataFrame({'Name':['Raju','Rani',None,'Anu'],'marks':[85,None,77,88]})
print("Nulls:\n",df.isnull().sum())
df['Name']=df['Name'].fillna('Unknown')
df['marks']=df['marks'].fillna(df['marks'].mean())
print("\n After Fill:\n",df)
```

```python
df=pd.DataFrame({'Name':['Raji','Rani','Ravi'],'marks':[85,92,77]})
print(df.sort_values(by='marks',ascending=False))
df['Result']=df['marks'].apply(lambda x: 'Pass' if x>80 else 'Fail')
print(df)
```

```python
df = pd.DataFrame({
    'Branch':['CSE','CSE','ECE','CSE','ECE'],
    'Marks':[85,92,77,88,95]
})

print(df.groupby('Branch').agg(
    Max_Marks=('Marks','max'),
    Mean_Marks=('Marks','mean'),
    Count=('Marks','count')
))
```
