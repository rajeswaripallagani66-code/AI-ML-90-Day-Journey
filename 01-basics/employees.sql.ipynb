```python
import sqlite3, pandas as pd
conn=sqlite3.connect('ml1.db')

df=pd.DataFrame({'id':[1,2,3,4],'name':['Ram','Sita','Arjun','Ravi'],'dept':['IT','HR','IT','Finance'],'salary':[60000,50000,80000,45000],'city':['Hyd','Hyd','Vizag','Hyd']               })
df.to_sql('employees',conn,if_exists='replace',index=False)

df_dept_info = pd.DataFrame({
    'dept':['IT','HR','Finance'],
    'manager':['Anil','Priya','Kiran']
})
df_dept_info.to_sql('DeptInfo', conn, if_exists='replace', index=False)

df=pd.read_sql_query("SELECT * FROM employees",conn)
print(df)
df_data=pd.read_sql_query("SELECT name,salary FROM employees",conn)
print(df_data)
df_from=pd.read_sql_query("SELECT * FROM employees WHERE city=='Hyd' ",conn)
print(df_from)
df_sal=pd.read_sql_query("SELECT *FROM employees WHERE salary>55000",conn)
print(df_sal)
#df_dep=pd.read_sql_query("SELECT dept, COUNT(*) WHERE employess GROUP BY dept",conn)
df_dep=pd.read_sql_query("SELECT dept, COUNT(*) as count FROM employees GROUP BY dept",conn)
print(df_dep)
#df_sum=pd.read_sql_query("SELECT city, salary(sum ) as total_salary FROM employees GROUP BY city",conn)
#print(df_sum)
df_salary=pd.read_sql_query("SELECT *FROM employees ORDER BY salary DESC LIMIT 2",conn)
print(df_salary)
# Now JOIN will work
print("--- JOIN Result ---")
df_join = pd.read_sql_query("""
    SELECT e.name, e.dept, d.manager
    FROM employees e
    JOIN DeptInfo d ON e.dept = d.dept
""", conn)
print(df_join)
```
